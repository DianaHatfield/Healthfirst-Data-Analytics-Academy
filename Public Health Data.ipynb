```python
group_column = "diabetesMed"
print(df[group_column].value_counts(dropna=False))
```

```python
feature = "time_in_hospital"
print(df[feature].head())
```

```python
df = df.replace("?", pd.NA)

df_clean = df.dropna(subset=[group_column, feature]).copy()
df_clean[feature] = pd.to_numeric(df_clean[feature], errors="coerce")
df_clean = df_clean.dropna(subset=[feature])

print(df_clean[[group_column, feature]].head())
print(df_clean.shape)
```

```python
group_yes = df_clean[df_clean[group_column] == "Yes"]
group_no = df_clean[df_clean[group_column] == "No"]

print("DiabetesMed = Yes:", group_yes.shape)
print("DiabetesMed = No:", group_no.shape)
```

```python
print("Descriptive statistics for DiabetesMed = Yes")
print(group_yes[feature].describe())

print("\nDescriptive statistics for DiabetesMed = No")
print(group_no[feature].describe())
```

```python
avg_yes = group_yes[feature].mean()
avg_no = group_no[feature].mean()

print("Average time_in_hospital for DiabetesMed = Yes:", avg_yes)
print("Average time_in_hospital for DiabetesMed = No:", avg_no)
```
