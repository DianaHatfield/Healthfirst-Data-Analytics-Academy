```python
def determine_winner(player1, player2):
    if player1 == player2:
        return "It's a tie!"
    
    if (player1 == "rock" and player2 == "scissors") or \
       (player1 == "scissors" and player2 == "paper") or \
       (player1 == "paper" and player2 == "rock"):
        return "Player 1 wins! Congratulations!"
    else:
        return "Player 2 wins! Congratulations!"


while True:
    print("\n--- Rock Paper Scissors ---")
    
    player1 = input("Player 1, enter rock, paper, or scissors: ").lower()
    player2 = input("Player 2, enter rock, paper, or scissors: ").lower()
    
    valid_choices = ["rock", "paper", "scissors"]
    
    if player1 not in valid_choices or player2 not in valid_choices:
        print("Invalid input! Please enter rock, paper, or scissors.")
        continue
    
    result = determine_winner(player1, player2)
    print(result)
    
    play_again = input("Do you want to play again? (yes/no): ").lower()
    if play_again != "yes":
        print("Thanks for playing!")
        break
```
