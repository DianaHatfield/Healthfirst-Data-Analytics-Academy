```python
import pandas as pd
```

```python
import seaborn as sns
```

```python
df=pd.read_csv('NYPD_Arrest_Data_(Year_to_Date)_20260506.csv')
```

```python
df
```

```python
df.columns
```

```python
df_clean = df.dropna(subset=["ARREST_PRECINCT", "ARREST_BORO", "OFNS_DESC"])
```

```python
df_clean.head()
```

```python
df_clean["ARREST_PRECINCT"].value_counts().head(10)
```

```python
precinct_1 = 14
precinct_2 = 40

comparison_df = df_clean[df_clean["ARREST_PRECINCT"].isin([precinct_1, precinct_2])]

comparison_df.head()
```

```python
offense_counts = comparison_df.groupby(["ARREST_PRECINCT", "OFNS_DESC"]).size().reset_index(name="arrest_count")

offense_counts.head()
```

```python
comparison_df.groupby("ARREST_PRECINCT").size()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# total arrests per precinct
precinct_totals = comparison_df.groupby("ARREST_PRECINCT").size().reset_index(name="count")

sns.barplot(data=precinct_totals, x="ARREST_PRECINCT", y="count")

plt.title("Total Arrests by Precinct")
plt.xlabel("Precinct")
plt.ylabel("Number of Arrests")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

sns.barplot(
    data=offense_counts,
    x="OFNS_DESC",
    y="arrest_count",
    hue="ARREST_PRECINCT"
)

plt.title("Arrests by Offense and Precinct")
plt.xlabel("Offense Description")
plt.ylabel("Arrest Count")

plt.xticks(rotation=45, ha="right")
plt.legend(title="Precinct")

plt.tight_layout()
plt.show()
```

```python
comparison_df.groupby("ARREST_PRECINCT")["AGE_GROUP"].value_counts()
```

```python
# remove null values first
clean_df = comparison_df[comparison_df["AGE_GROUP"] != "(null)"]

# then group
age_data = clean_df.groupby("ARREST_PRECINCT")["AGE_GROUP"].value_counts().reset_index(name="count")

age_data
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

sns.barplot(
    data=age_data,
    x="AGE_GROUP",
    y="count",
    hue="ARREST_PRECINCT"
)

plt.title("Age Group Distribution by Precinct")
plt.xlabel("Age Group")
plt.ylabel("Number of Arrests")

plt.xticks(rotation=45)
plt.legend(title="Precinct")

plt.tight_layout()
plt.show()
```

```python
top_offenses = comparison_df["OFNS_DESC"].value_counts().head(10)

plt.figure(figsize=(10,5))
top_offenses.plot(kind="bar")
plt.title("Top 10 Arrest Offenses")
plt.xlabel("Offense Description")
plt.ylabel("Number of Arrests")
plt.xticks(rotation=45, ha="right")
plt.show()
```
